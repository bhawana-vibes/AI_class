```python
#melt and pivot
df = pd.DataFrame({
    "country": ["usa" , "usa" , "india" , "india"],
    "year": [2020 , 2021 , 2020 , 2021],
    "sales": [100 , 110 , 90 , 120],
    "profit": [20 , 21 , 25 , 22],
    #"tax": [10,20,30,40]
})

melted_df = df.melt(
    id_vars = ["country" , "year"],
    value_vars = ["sales" , "profit" ],
    var_name = "metrics",
    value_name = "value"
)

melted_df.pivot(
    index=["country" , "year"] ,
    columns="metrics",
    values="value"
)
```

```python
#mering and joining
df_customers = pd.DataFrame({
    "customer_id" : [1,2,3,4],
    "name" : ["aa" , "bb" , "cc" ,"dd"]
})

df_orders = pd.DataFrame({
    "order_id" : [101 ,102 ,103 ,104],
    "customer_id" : [2,4,1,5],
    "amount" : [100,234,345,765]
})

pd.merge(df_customers , df_orders , on="customer_id")#inner join
pd.merge(df_customers , df_orders , on="customer_id" , how = "left")#left join
pd.merge(df_customers , df_orders , on="customer_id" , how = "right")#rightjoin
pd.merge(df_customers , df_orders , on="customer_id" , how = "outer")#outer join
```
