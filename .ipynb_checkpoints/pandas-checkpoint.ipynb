```python
#dataframe
info ={
    "name" : ["aayu" , "shipi" , "anj" , "kb"],
    "cgpa" : [9.5 , 7.5 , 5.6 ,7.8],
    "age" : [21, 22,23,25]
}

df = pd.DataFrame(info)
print(df)
print(df.index)
print(df.columns)
```
