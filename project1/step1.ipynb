```python
def check_book(new_price, old_price, syllabus):
    saving = new_price - old_price
    if saving > 50 and syllabus >= 90:
        return "Buy Old Book", saving
    return "Buy New Book", saving
all_books = []
total_savings = 0
for i in range(5):
 book_name = input("enter book name:")
 new_book = int(input("enter the new book price:"))
 old_book = int(input("enter the old book price:"))
 syllabus_book = 95
 decision, saving = check_book(new_book, old_book, syllabus_book)
 saving_percent = (saving / new_book)*100
 if decision == "Buy Old Book":
        total_savings += saving
 print(f"{book_name} on saving: Rs. {saving} ({round(saving_percent)}% saved!)")
 all_books.append({"name": book_name , "saving" : saving})
print(f"\nTotal Savings for 5 books: Rs. {total_savings}")

    
```
