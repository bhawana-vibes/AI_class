```python
book_name = input("enter book name:")
new_book = int(input("new book ka price daalo:"))
old_book = int(input("old book ka price daalo:"))
saving = new_book - old_book
prt = (saving / new_book)*100
print(f"{book_name} par saving: Rs. {saving} ({round(prt)}% bacha!)")
```
